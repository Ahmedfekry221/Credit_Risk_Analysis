```python
# importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/German_credit_Data.csv')
# show the first 5 rows of the dataset
display(df.head())
```

```python
print("\n--- Missing Values ---")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='class', hue='class', data=df, palette='Set1')
plt.title('Target Variable Distribution (Credit Risk)')
plt.show()
```

```python
# 3. Explore the relationship between (Credit Amount) and (Credit Risk)
# Do people taking larger loans default more often?
plt.figure(figsize=(8, 5))
sns.boxplot(x='class', hue='class', y='credit_amount', data=df, palette='Set2')
plt.title('Credit Amount vs. Credit Risk')
plt.show()
```

```python
# 4. Explore the relationship between (Loan Purpose) and (Credit Risk)
plt.figure(figsize=(12, 6))
sns.countplot(x='purpose', hue='class', data=df, palette='Set1')
plt.title('Loan Purpose vs. Credit Risk')
plt.xticks(rotation=45)
plt.show()
```

```python
# 5. Correlation matrix for numerical columns (Correlation Heatmap)
plt.figure(figsize=(8, 6))

# Extract only numerical columns (e.g., age, duration, and credit amount)
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
correlation_matrix = df[numerical_cols].corr()

# Plot the heatmap
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```
